# Load and Explore the Dataset

```python
import warnings
from pathlib import Path

import pandas as pd

warnings.filterwarnings("ignore")


# Load all cryptocurrency data
data_path = Path("data")
crypto_data = {}

for csv_file in sorted(data_path.glob("*.csv")):
    crypto_name = csv_file.stem
    df = pd.read_csv(csv_file)
    df["Date"] = pd.to_datetime(df["Date"])
    df = df.sort_values("Date")
    crypto_data[crypto_name] = df

print(f"Loaded {len(crypto_data)} cryptocurrencies")
print(f"\nData date range: {df['Date'].min()} to {df['Date'].max()}")
print(f"Total records per crypto: {len(df)}")

# Create summary statistics
summary_stats = {}
for crypto, df in crypto_data.items():
    summary_stats[crypto] = {
        "Min Price": df["Close"].min(),
        "Max Price": df["Close"].max(),
        "Mean Price": df["Close"].mean(),
        "Std Dev": df["Close"].std(),
        "Range": df["Close"].max() - df["Close"].min(),
        "CV (Coefficient of Variation)": df["Close"].std()
        / df["Close"].mean(),
    }

summary_df = pd.DataFrame(summary_stats).T
summary_df = summary_df.sort_values("Mean Price", ascending=False)
print("\nSummary Statistics by Cryptocurrency:")
print(summary_df.to_string())
```
